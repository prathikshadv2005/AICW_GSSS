```python
class student:
    def __init__(self, name, place):
        self.name = name
        self.place = place

    def show(self):
        return f'The name is {self.name} and you are from {self.place}'

p1 = student('abc','mysore')
p1.show()
```

```python
p1.name
```

```python
p1.place
```

```python
class Student:
    def __init__(std, usn, name, place):
        std.usn = usn
        std.name= name
        std.place = place
    def show(std):
        print(f'USN of {std.usn} having the name {std.name} is from {std.place}')
    def update(std, usn, place):
        std.usn = usn
        std.place = place
```

```python
stud = Student(114, 'abc', 'mys')
stud.name
```

```python
stud.show()
```

```python
stud.update(19,'blore')
```

```python
stud.show()
```
